# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?

### Answer
### 1. Partition Function

The canonical partition function is given by:
$$Z(T) = \sum_{j=0}^{\infty} e^{-E_j / (k_B T)}$$

Substituting the energy eigenvalues for a quantum harmonic oscillator, $E_j = \hbar\omega\left(j + \frac{1}{2}\right)$, gives:
$$Z = \sum_{j=0}^{\infty} e^{-\frac{\hbar\omega\left(j+\frac{1}{2}\right)}{k_B T}}$$

We can separate the exponential into two factors:
$$Z = e^{-\frac{\hbar\omega}{2k_B T}} \sum_{j=0}^{\infty} e^{-\frac{\hbar\omega j}{k_B T}}$$

The summation is a geometric series with a common ratio $r = e^{-\frac{\hbar\omega}{k_B T}}$, where $0 < r < 1$. Using the geometric series sum formula $\sum_{j=0}^{\infty} r^j = \frac{1}{1-r}$, we get:

$$Z(T) = \frac{e^{-\frac{\hbar\omega}{2k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}}$$

---

### 2. Thermodynamic Potentials

<u>(a) Helmholtz Free Energy</u>
The Helmholtz free energy is defined as $F = -k_B T \ln Z$. Taking the natural logarithm of the partition function:
$$\ln Z = -\frac{\hbar\omega}{2k_B T} - \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)$$

Therefore:
$$F(T) = \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)$$

<u> (b) Internal Energy</u>
The internal energy $U$ can be found using the relation $U = -\frac{\partial \ln Z}{\partial \beta}$, where $\beta = \frac{1}{k_B T}$. Expressing $\ln Z$ in terms of $\beta$:
$$\ln Z = -\frac{\beta \hbar\omega}{2} - \ln\left(1 - e^{-\beta \hbar\omega}\right)$$

Differentiating with respect to $\beta$:
$$\frac{\partial \ln Z}{\partial \beta} = -\frac{\hbar\omega}{2} - \frac{\hbar\omega}{e^{\beta \hbar\omega} - 1}$$

Hence, the internal energy is:
$$U(T) = \hbar\omega \left( \frac{1}{2} + \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1} \right)$$

> **Note:** The first term represents the quantum **zero-point energy**, while the second term accounts for the **thermal excitation energy**.

---

### 3. Entropy

The entropy $S$ is obtained by taking the negative temperature derivative of the Helmholtz free energy, $S = -\left(\frac{\partial F}{\partial T}\right)$.

Differentiating $F(T)$ yields:
$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{T} \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1}$$

<u>Verification of $U = F + TS$ </u><br>
Substituting our expressions for $F$ and $S$:
$$F + TS = \left[\frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)\right] + T \left[-k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{T} \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1}\right]$$

The logarithmic terms cancel out completely, leaving:
$$F + TS = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_B T}} - 1} = U$$

Thus, the thermodynamic identity **$U = F + TS$** is verified.

---

### 4. Heat Capacity

The constant-volume heat capacity is defined as $C_V = \left(\frac{\partial U}{\partial T}\right)$. Differentiating the internal energy $U(T)$ with respect to $T$ gives:
$$C_V = k_B \left( \frac{\hbar\omega}{k_B T} \right)^2 \frac{e^{\frac{\hbar\omega}{k_B T}}}{\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^2}$$

<u>Hyperbolic Form</u>
We can rewrite this in a cleaner form using the hyperbolic sine identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$. Let $x = \frac{\hbar\omega}{2k_B T}$. The exponential term can be rearranged as:
$$\frac{e^{\frac{\hbar\omega}{k_B T}}}{\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^2} = \frac{1}{4\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)}$$

Substituting this back into the heat capacity equation yields:
$$C_V = k_B \left( \frac{\hbar\omega}{2k_B T} \right)^2 \frac{1}{\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)}$$

---

### 5. Asymptotic Behaviour

In the **high-temperature limit** where $k_B T \gg \hbar\omega$, the exponential term can be approximated via Taylor expansion:
$$e^{\frac{\hbar\omega}{k_B T}} \approx 1 + \frac{\hbar\omega}{k_B T}$$

Substituting this linear approximation into our heat capacity expression leads to:
$$\lim_{T \to \infty} C_V = k_B$$

<u>Discussion</u>
At high temperatures, the thermal energy scale is significantly larger than the quantum spacing ($\hbar\omega$) between adjacent energy levels. Consequently, the quantum features blur out, and the system behaves like a classical harmonic oscillator.

According to the **classical equipartition theorem**, a 1D harmonic oscillator possesses two quadratic degrees of freedom (kinetic energy and potential energy). Each degree of freedom contributes $\frac{1}{2}k_B T$ to the internal energy:
$$U \approx k_B T \implies C_V = \frac{dU}{dT} = k_B$$

This confirms that the quantum harmonic oscillator smoothly transitions to match the classical equipartition theorem prediction at high temperatures.

# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

### Answer The Maximum Entropy Distribution under Mean and Variance Constraints

Consider a continuous random variable $X \in (-\infty, \infty)$. The known constraints are:
$$E[X] = \mu$$
$$E[(X-\mu)^2] = \sigma^2$$

We wish to determine the probability density function $p(x)$ that maximizes the Shannon differential entropy:
$$H[p] = -\int_{-\infty}^{\infty} p(x) \ln p(x) \, dx$$

subject to the constraints:

* **Normalization:** $\int_{-\infty}^{\infty} p(x) \, dx = 1$
* **Mean:** $\int_{-\infty}^{\infty} x p(x) \, dx = \mu$
* **Variance:** $\int_{-\infty}^{\infty} (x-\mu)^2 p(x) \, dx = \sigma^2$

<u>Constructing the Lagrangian</u>

Introduce the Lagrange multipliers $\gamma, \lambda_1$, and $\lambda_2$. The functional becomes:
$$L[p] = -\int_{-\infty}^{\infty} p(x) \ln p(x) \, dx - \gamma \left( \int_{-\infty}^{\infty} p(x) \, dx - 1 \right) - \lambda_1 \left( \int_{-\infty}^{\infty} x p(x) \, dx - \mu \right) - \lambda_2 \left( \int_{-\infty}^{\infty} (x-\mu)^2 p(x) \, dx - \sigma^2 \right)$$

<u>Taking the Functional Derivative</u>

Setting the functional derivative with respect to $p(x)$ to zero, $\frac{\delta L}{\delta p(x)} = 0$, gives:
$$-\ln p(x) - 1 - \gamma - \lambda_1 x - \lambda_2 (x-\mu)^2 = 0$$

Hence:
$$\ln p(x) = -(1 + \gamma) - \lambda_1 x - \lambda_2 (x-\mu)^2$$

Exponentiating both sides:
$$p(x) = e^{-(1+\gamma)} e^{-\lambda_1 x} e^{-\lambda_2 (x-\mu)^2}$$

<u>Determining the Multipliers</u>

Since the mean is fixed at $\mu$, symmetry across the coordinate translation requires:
$$\lambda_1 = 0$$

Thus, the expression simplifies to:
$$p(x) = \frac{1}{Z} e^{-\lambda_2 (x-\mu)^2}$$

where the normalization constant is $Z = e^{1+\gamma}$.

<u>Computing the Normalization Constant</u>

Normalization requires:
$$1 = \int_{-\infty}^{\infty} \frac{1}{Z} e^{-\lambda_2 (x-\mu)^2} \, dx$$

Using the standard Gaussian integral identity $\int_{-\infty}^{\infty} e^{-a(x-\mu)^2} \, dx = \sqrt{\frac{\pi}{a}}$, we obtain:
$$Z = \sqrt{\frac{\pi}{\lambda_2}}$$

<u>Using the Variance Constraint</u>

The variance of a Gaussian distribution satisfies:
$$\sigma^2 = \frac{1}{2\lambda_2}$$

Therefore:
$$\lambda_2 = \frac{1}{2\sigma^2}$$

Substituting $\lambda_2$ back into the expression for $Z$ yields:
$$Z = \sqrt{2\pi\sigma^2}$$

<u>Final Maximum Entropy Distribution</u>

Hence, substituting the solved parameters back into $p(x)$:
$$p(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x-\mu)^2}{2\sigma^2} \right)$$

Therefore:
$$X \sim N(\mu, \sigma^2)$$

<u>Conclusion</u>
The unique probability density function that maximizes Shannon differential entropy among all continuous distributions with a specified mean and variance is the **Gaussian (Normal) distribution**:
$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x-\mu)^2}{2\sigma^2} \right)$$

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.



### Answer - Multiclass Logistic Regression: Automatic Classification of Emails
Let $p(y \mid x)$ be the conditional probability that an email with feature vector $x$ is routed to category $y$. The conditional entropy is defined as:
$$H(Y \mid X) = -\sum_x P(x) \sum_y p(y \mid x) \ln p(y \mid x)$$

The Maximum Entropy principle chooses the conditional distribution that maximizes $H(Y \mid X)$ subject to the observed constraints:
* $E[x_1 I(Y=1)] = 0.05$
* $E[x_2 I(Y=1)] = 0.15$
* $E[x_3 I(Y=1)] = 0.16$

Expanding these expectations yields:
$$\sum_x P(x) x_1 p(Y=1 \mid x) = 0.05$$
$$\sum_x P(x) x_2 p(Y=1 \mid x) = 0.15$$
$$\sum_x P(x) x_3 p(Y=1 \mid x) = 0.16$$



<u>Lagrangian</u>

Introducing the Lagrange multipliers $\lambda_1, \lambda_2, \lambda_3$, the Lagrangian $L$ is constructed as:
$$L = -H(Y \mid X) + \lambda_1 (E[f_1] - 0.05) + \lambda_2 (E[f_2] - 0.15) + \lambda_3 (E[f_3] - 0.16)$$

where the feature functions are defined as:
* $f_1 = x_1 I(Y=1)$
* $f_2 = x_2 I(Y=1)$
* $f_3 = x_3 I(Y=1)$



<u>Maximum Entropy Solution</u>

Setting the functional derivatives equal to zero yields the exponential-family form:
$$p(y \mid x) = \frac{\exp\left( \sum_i \lambda_i f_i(x,y) \right)}{Z(x)}$$

For the Technical category ($Y=1$):
$$p(Y=1 \mid x) = \frac{\exp(\lambda_1 x_1 + \lambda_2 x_2 + \lambda_3 x_3)}{Z(x)}$$

Generalizing for all classes, we obtain:
$$p(y \mid x) = \frac{\exp(w_y^T x)}{\sum_k \exp(w_k^T x)}$$


<u>Final Logistic Regression Form</u>

Therefore, the Maximum Entropy classifier is exactly the multiclass softmax model:
$$p(y \mid x) = \frac{\exp(w_y^T x)}{\sum_k \exp(w_k^T x)}$$

which is equivalent to the standard multiclass logistic regression model.

> **Observation:** Because $E[x_3 I(Y=1)] = 0.16$ is the strongest constraint, it follows that $\lambda_3 > \lambda_2 > \lambda_1$. Thus, the presence of bug-related keywords ($x_3$) contributes most strongly toward classification into the Technical department.

###Answer- Asset Pricing and Portfolio Optimization

Entropy Functional

Let $S = S_T$ denote the future asset price. The differential entropy is:
$$H[p] = -\int_0^\infty p(S) \ln p(S) \, dS$$

We maximize $H[p]$ subject to the specific market constraints.



<u>Constraints</u>

* **Normalization:** $\int_0^\infty p(S) \, dS = 1$
* **Expected Price:** $\int_0^\infty S p(S) \, dS = 102$
* **Second Moment:** $\int_0^\infty S^2 p(S) \, dS = 10600$
* **Logarithmic Constraint:** $\int_0^\infty \ln(S) p(S) \, dS = 4.60$

<u>Construct the Lagrangian</u>

Introducing multipliers $\alpha, \beta, \gamma, \delta$, the Lagrangian functional is:
$$L = -\int_0^\infty p(S) \ln p(S) \, dS - \alpha \left( \int_0^\infty p(S) \, dS - 1 \right) - \beta \left( \int_0^\infty S p(S) \, dS - 102 \right) - \gamma \left( \int_0^\infty S^2 p(S) \, dS - 10600 \right) - \delta \left( \int_0^\infty \ln(S) p(S) \, dS - 4.60 \right)$$

<u>Stationary Condition</u>

Setting the functional derivative to zero, $\frac{\delta L}{\delta p(S)} = 0$, gives:
$$-\ln p(S) - 1 - \alpha - \beta S - \gamma S^2 - \delta \ln S = 0$$

Hence:
$$\ln p(S) = -(1 + \alpha) - \beta S - \gamma S^2 - \delta \ln S$$

Exponentiating both sides yields:
$$p(S) = e^{-(1+\alpha)} S^{-\delta} e^{-\beta S - \gamma S^2}$$

<u>Maximum Entropy Distribution</u>

The normalization constant $Z$ is given by:
$$Z = \int_0^\infty S^{-\delta} e^{-\beta S - \gamma S^2} \, dS$$

Therefore, for $S_T > 0$, the probability distribution is:
$$p(S_T) = \frac{1}{Z} S_T^{-\delta} \exp\left( -\beta S_T - \gamma S_T^2 \right)$$

<u>Interpretation</u>

The resulting Maximum Entropy distribution is **not** Gaussian.

The additional constraint $E[\ln S_T] = 4.60$ introduces the power-law factor $S_T^{-\delta}$, which produces skewness and heavier tails than a standard normal distribution.

Without the logarithmic constraint ($\delta = 0$), the solution would reduce to the exponential-quadratic form $p(S_T) \propto e^{-\beta S_T - \gamma S_T^2}$, which is closely related to a truncated Gaussian on $S_T > 0$.

Thus, the market-implied downside risk modifies the distribution into:
$$p(S_T) = \frac{1}{Z} S_T^{-\delta} e^{-\beta S_T - \gamma S_T^2}$$

This represents the least-biased distribution consistent with all observed market information.
