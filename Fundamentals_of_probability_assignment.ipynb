# Q1) The Probability of Shooting at a Target

A point is chosen at random from a disk of radius $10$. Let $A$ be the event that the point lies within $1$ unit of the boundary.

1. Model the experiment assuming the point is uniformly distributed with respect to area, and compute $P(A)$.
2. Model the experiment assuming the distance to the point from the center is chosen uniformly from $[0,10]$, and the direction is chosen independently and uniformly from $[0,2\pi)$ and compute $P(A)$.
3. Explain why the two answers differ.


#### Answer

1. Let the radius be $R = 10$.

Total area:
$$
A_{\text{total}} = \pi R^2 = 100\pi
$$

Let event $A$: point lies within radius $a$.

Then:
$$
P(A) = \frac{\pi a^2}{\pi 10^2}
$$

$$
P(A) = \frac{a^2}{100}
$$

<br><br>



2. $r \sim \text{Uniform}(0,10)$  
$\theta \sim \text{Uniform}(0,2\pi)$

PDF of $r$:
$$
f_r(r) = \frac{1}{10}, \quad 0 \le r \le 10
$$

Then:
$$
P(A) = P(r \le a)
$$

$$
P(A) = \int_0^a \frac{1}{10} \, dr
$$

$$
P(A) = \frac{a}{10}
$$



3. Reasons that the answers differ are;

- Model 1 is uniform over area → probability depends on $r^2$
- Model 2 is uniform in $r$ → probability depends on $r$

Correct area element:
$$
dA = r \, dr \, d\theta
$$

So uniform in $r$ is NOT the same as uniform in area.

# Q2) Bertrand's Paradox

A chord of a circle is chosen “at random.” What is the probability that its length exceeds the side length of the inscribed equilateral triangle?

Show that different mathematically natural models of “random chord” lead to different answers.

#### Answer

A chord is longer than the side of the inscribed equilateral triangle **iff**
its distance from the center satisfies:

$$
d < \frac{R}{2}
$$

## Model 1: Midpoint uniformly distributed over the area

Midpoint is uniformly distributed in the disk.

$$
P = \frac{\pi (R/2)^2}{\pi R^2}
$$

$$
P = \frac{1}{4}
$$

<br><br>

## Model 2: Uniform point along a radius

Choose a random radius and a point uniformly along it.  
Then $d \sim \text{Uniform}(0, R)$.

$$
P = P\left(d < \frac{R}{2}\right)
$$

$$
P = \frac{R/2}{R}
$$

$$
P = \frac{1}{2}
$$

<br><br>

## Model 3: Random endpoints on circumference

Pick two points uniformly on the circle.

Chord length:
$$
L = 2R \sin\left(\frac{\theta}{2}\right)
$$

Condition:
$$
2R \sin\left(\frac{\theta}{2}\right) > \sqrt{3}R
$$

$$
\sin\left(\frac{\theta}{2}\right) > \frac{\sqrt{3}}{2}
$$

$$
\theta > \frac{2\pi}{3}
$$

Since $\theta \sim \text{Uniform}(0,\pi)$:

$$
P = \frac{\pi - \frac{2\pi}{3}}{\pi}
$$

$$
P = \frac{1}{3}
$$

<br><br>

## Why the answers differ

Each model defines “random chord” differently:

- Model 1 → uniform over area  
- Model 2 → uniform in radial distance  
- Model 3 → uniform in angle  

These produce different probability distributions over chords.

**Conclusion:** The problem is ambiguous unless the randomization method is specified.

#Q8) Probability of Being a Girl?

A family has two children. Assume each child is equally likely to be a boy or a girl, independently of the other.

What is the probability that both children are girls, given that **at least one** of them is a girl?

#### Answer
A family has two children. Each child is equally likely to be a boy (B) or a girl (G), independently.

<br>Sample Space

All equally likely outcomes:

$$
S = \{BB, BG, GB, GG\}
$$

Each outcome has probability:

$$
P = \frac{1}{4}
$$

<br>Given Condition

“At least one child is a girl”

This removes the outcome $BB$.

Remaining outcomes:

$$
\{BG, GB, GG\}
$$

<br>Required Probability

We want:

$$
P(\text{both are girls} \mid \text{at least one is a girl})
$$

Only one favorable case:

$$
GG
$$

Total possible cases under condition:

$$
3
$$

<br>Final Answer

$$
P = \frac{1}{3}
$$

# Q9) Information Available in a Discrete Random Variable

A factory inspects a product and records its quality using two different coding systems.

Each product falls into exactly one of the following four categories:

* $G_1$: good quality, produced on Machine 1
* $G_2$: good quality, produced on Machine 2
* $D_1$: defective, produced on Machine 1
* $D_2$: defective, produced on Machine 2

Assume the sample space is

$$
\Omega=\{G_1,G_2,D_1,D_2\},
$$

with probabilities

$$
P(\{G_1\})=0.50,\qquad P(\{G_2\})=0.20,\qquad P(\{D_1\})=0.10,\qquad P(\{D_2\})=0.20.
$$

Define two discrete random variables on the same probability space:

* $X$ records only whether the item is good or defective:
  $$
  X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1.
  $$

* $Y$ records both quality and machine:
  $$
  Y(G_1)=1,\quad Y(G_2)=2,\quad Y(D_1)=3,\quad Y(D_2)=4.
  $$

Answer the following.

1. Write down the probability space $(\Omega,\mathfrak F,P)$, where $\mathfrak F=\mathfrak P(\Omega)$.

2. Compute the $\sigma$-algebra generated by $X$, namely $\sigma(X)$.

3. Compute the $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$.

4. Show that $\sigma(X)\subseteq \sigma(Y)$, and explain why $Y$ contains more information than $X$.

5. Find the marginal distribution of $X$.

6. Find the marginal distribution of $Y$.

7. Compute the joint probabilities
   $$
   P(X=0,Y=1),\quad P(X=0,Y=2),\quad P(X=1,Y=3),\quad P(X=1,Y=4).
   $$

8. Compute the conditional probabilities
   $$
   P(X=1\mid Y=3),\qquad P(X=1\mid Y=4),\qquad P(Y=3\mid X=1),\qquad P(Y=4\mid X=1).
   $$

9. Interpret the difference between conditioning on $X$ and conditioning on $Y$.

10. Explain, in words, what information is available in $\sigma(X)$ but not in $\sigma(Y)$, and what information is available in $\sigma(Y)$ but not in $\sigma(X)$.

11. Give a Python based visualization of the model.

12. Find the marginal, joint, and conditional Shannon Entropy of $X$ and $Y$.

#### Answer
1. Probability Space

Sample space:
$$
\Omega = \{G_1, G_2, D_1, D_2\}
$$

Sigma-algebra:
$$
\mathfrak{F} = \mathfrak{P}(\Omega)
$$

Probability measure:
$$
P(G_1)=0.50,\quad P(G_2)=0.20,\quad P(D_1)=0.10,\quad P(D_2)=0.20
$$

<br>2. Random Variables

<br> Variable $X$ (quality only)
$$
X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1
$$

<br> Variable $Y$ (quality + machine)
$$
Y(G_1)=1,\quad Y(G_2)=2,\quad Y(D_1)=3,\quad Y(D_2)=4
$$

<br>3. Sigma-algebra generated by $X$

Partition induced by $X$:
- $\{G_1, G_2\}$ (good)
- $\{D_1, D_2\}$ (defective)

So:
$$
\sigma(X) = \{\emptyset, \Omega, \{G_1,G_2\}, \{D_1,D_2\}\}
$$

<br>4. Sigma-algebra generated by $Y$

Each outcome is distinguishable, so:
$$
\sigma(Y) = \mathfrak{P}(\Omega)
$$

<br>5. Relation between $\sigma(X)$ and $\sigma(Y)$

$$
\sigma(X) \subseteq \sigma(Y)
$$

Reason:
- $X$ only tells **good/defective**
- $Y$ tells **good/defective + machine**

So $Y$ contains strictly more information.

<br> 6. Marginal Distribution of $X$

$$
P(X=0) = P(G_1)+P(G_2) = 0.50+0.20 = 0.70
$$

$$
P(X=1) = P(D_1)+P(D_2) = 0.10+0.20 = 0.30
$$

<br> 7. Marginal Distribution of $Y$

$$
P(Y=1)=0.50,\quad P(Y=2)=0.20,\quad P(Y=3)=0.10,\quad P(Y=4)=0.20
$$

<br> 8. Joint Probabilities

$$
P(X=0,Y=1)=0.50
$$
$$
P(X=0,Y=2)=0.20
$$
$$
P(X=1,Y=3)=0.10
$$
$$
P(X=1,Y=4)=0.20
$$

<br>9. Conditional Probabilities

$$
P(X=1 \mid Y=3)=1,\quad P(X=1 \mid Y=4)=1
$$

$$
P(Y=3 \mid X=1)=\frac{0.10}{0.30}=\frac{1}{3}
$$

$$
P(Y=4 \mid X=1)=\frac{0.20}{0.30}=\frac{2}{3}
$$

<br>10. Interpretation

- Conditioning on $Y$ → exact outcome known → no uncertainty in $X$
- Conditioning on $X$ → only partial info → uncertainty remains in $Y$

<br>11. Information Comparison

- $\sigma(X)$: only distinguishes **good vs defective**
- $\sigma(Y)$: distinguishes **all four categories**

Thus:
- Nothing in $\sigma(X)$ is missing from $\sigma(Y)$
- But $\sigma(Y)$ contains **machine-level detail** not in $\sigma(X)$

<br>12. Entropy

<br>Entropy of $X$
$$
H(X) = -[0.7\log_2(0.7) + 0.3\log_2(0.3)]
$$

<br>Entropy of $Y$
$$
H(Y) = -\sum p_i \log_2 p_i
$$

<br>Conditional Entropy
$$
H(X|Y)=0
$$

$$
H(Y|X)=0.7\cdot H\left(\frac{0.5}{0.7},\frac{0.2}{0.7}\right)
+ 0.3\cdot H\left(\frac{0.1}{0.3},\frac{0.2}{0.3}\right)
$$
