#Q3) Transformer Failiures

In a certain electricity distribution district, the number of emergency transformer failures reported in a single day is modeled as a discrete random variable $X$. Assume that $X$ takes values in the set $\{0,1,2,3\}$, with probabilities

$$
P(X=0)=0.50,\qquad P(X=1)=0.30,\qquad P(X=2)=0.15,\qquad P(X=3)=0.05.
$$

1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.
2. Define $X$ as a random variable on this space.
3. Determine the probability mass function of $X$.
4. Compute the expectation $E[X]$.
5. Compute the variance $\mathrm{Var}(X)$.
6. Find $P(X\ge 1)$ and $P(X\ge 2)$.
7. Briefly explain the practical meaning of the expectation and variance in this setting.



### Answer
Let $X$ be the number of emergency transformer failures reported in a single day.

The possible values are:
$$
\{0,1,2,3\}
$$

with probabilities:
$$
P(X=0)=0.50
$$

$$
P(X=1)=0.30
$$

$$
P(X=2)=0.15
$$

$$
P(X=3)=0.05
$$

1. Probability Space

Define:
$$
\Omega = \{0,1,2,3\}
$$

Let:
$$
\mathfrak{F} = \mathcal{P}(\Omega)
$$

(the power set of $\Omega$)

Probability measure:
$$
P(\{0\})=0.50
$$

$$
P(\{1\})=0.30
$$

$$
P(\{2\})=0.15
$$

$$
P(\{3\})=0.05
$$

2. Definition of Random Variable

Define:
$$
X:\Omega \to \mathbb{R}
$$

by:
$$
X(\omega)=\omega
$$

Thus, $X$ records the number of transformer failures in a day.

3. Probability Mass Function (PMF)

The PMF of $X$ is:

$$
p_X(x)=P(X=x)
$$

Therefore,

$$
p_X(0)=0.50
$$

$$
p_X(1)=0.30
$$

$$
p_X(2)=0.15
$$

$$
p_X(3)=0.05
$$

4. Expectation of $X$

Using:
$$
E[X]=\sum xP(X=x)
$$

we get:

$$
E[X]
=0(0.50)+1(0.30)+2(0.15)+3(0.05)
$$

$$
E[X]
=0+0.30+0.30+0.15
$$

$$
E[X]=0.75
$$

5. Variance of $X$

First compute:
$$
E[X^2]
=\sum x^2P(X=x)
$$

$$
E[X^2]
=0^2(0.50)+1^2(0.30)+2^2(0.15)+3^2(0.05)
$$

$$
=0+0.30+0.60+0.45
$$

$$
E[X^2]=1.35
$$

Variance:
$$
\mathrm{Var}(X)=E[X^2]-(E[X])^2
$$

$$
\mathrm{Var}(X)=1.35-(0.75)^2
$$

$$
\mathrm{Var}(X)=1.35-0.5625
$$

$$
\mathrm{Var}(X)=0.7875
$$
6. Probabilities

Probability of at least one failure

$$
P(X\ge1)=1-P(X=0)
$$

$$
P(X\ge1)=1-0.50
$$

$$
P(X\ge1)=0.50
$$

Probability of at least two failures

$$
P(X\ge2)=P(X=2)+P(X=3)
$$

$$
P(X\ge2)=0.15+0.05
$$

$$
P(X\ge2)=0.20
$$

7. Practical Interpretation

Expectation

$$
E[X]=0.75
$$

This means that, on average, the district experiences about 0.75 emergency transformer failures per day.

Variance

$$
\mathrm{Var}(X)=0.7875
$$

The variance measures how much the daily number of failures fluctuates around the average value.A larger variance would indicate more unpredictability in the number of daily failures.

# Q4) Probability of Defective Items

In a manufacturing process, the proportion of defective items in a large batch produced on a given day is modeled by a continuous random variable $X$. Since $X$ represents a proportion, it takes values in the interval $[0,1]$. Suppose that $X$ has a Beta distribution with parameters $\alpha=2$ and $\beta=18$, so that its density is

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x^{1}(1-x)^{17}, & 0<x<1,\\\
0, & \text{otherwise},
\end{cases}
$$

where $B(2,18)$ is the Beta function.

Answer the following:

1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.

2. Define $X$ as a continuous random variable on this space.

3. Write down the probability density function of $X$.

4. Verify that $f$ is a valid probability density function.

5. Compute the expectation $E[X]$.

6. Compute the variance $\mathrm{Var}(X)$.

7. Find the probability that the defective proportion is at most $0.10$, that is, compute

$$
P(X\le 0.10).
$$

You may leave your answer in Beta integral form.

8. Find the probability that the defective proportion lies between $0.05$ and $0.15$, that is, compute

$$
P(0.05\le X\le 0.15).
$$

You may leave your answer in integral form.

9. Briefly explain why a Beta distribution is a reasonable model for this situation.

10. Interpret the expectation and variance in the context of production quality.


### Answer
Let $X$ denote the proportion of defective items in a production batch.

Since $X$ is a proportion:
$$
0 \le X \le 1
$$

Assume:
$$
X \sim \text{Beta}(2,18)
$$

with density
$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1 \\
0, & \text{otherwise}
\end{cases}
$$

1. Probability Space

Define:
$$
\Omega=[0,1]
$$

Let:
$$
\mathfrak{F}=\mathfrak{B}([0,1])
$$

where $\mathfrak{B}([0,1])$ is the Borel sigma-algebra on $[0,1]$.

The probability measure $P$ is induced by the density function $f(x)$.

2. Definition of Random Variable

Define:
$$
X:\Omega \to \mathbb{R}
$$

by:
$$
X(\omega)=\omega
$$

Thus, $X$ represents the defective proportion in the batch.

3. Probability Density Function

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1 \\
0, & \text{otherwise}
\end{cases}
$$

4. Verification that $f$ is a Valid PDF

A valid probability density function must satisfy:

(i) Non-negativity
$$
f(x)\ge0
$$

for all $x$.

Since:
$$
x \ge 0
\quad \text{and} \quad
(1-x)^{17}\ge0
$$

we have:
$$
f(x)\ge0
$$

(ii) Total probability equals 1

We verify:
$$
\int_{-\infty}^{\infty} f(x)\,dx =1
$$

Since the density is nonzero only on $(0,1)$:

$$
\int_0^1 \frac{1}{B(2,18)}x(1-x)^{17}dx
$$

Using the definition of the Beta function:

$$
B(2,18)=\int_0^1 x(1-x)^{17}dx
$$

therefore:

$$
\int_0^1 f(x)\,dx=1
$$

Hence, $f$ is a valid probability density function.

5. Expectation of $X$

For a Beta distribution:
$$
E[X]=\frac{\alpha}{\alpha+\beta}
$$

Here:
$$
\alpha=2,\quad \beta=18
$$

Thus:
$$
E[X]=\frac{2}{20}
$$

$$
E[X]=0.10
$$

6. Variance of $X$

For a Beta distribution:
$$
\mathrm{Var}(X)=
\frac{\alpha\beta}
{(\alpha+\beta)^2(\alpha+\beta+1)}
$$

Substituting values:

$$
\mathrm{Var}(X)=
\frac{(2)(18)}
{(20)^2(21)}
$$

$$
=
\frac{36}{8400}
$$

$$
\mathrm{Var}(X)=0.004286
$$

(approximately)

7. Probability that $X \le 0.10$

$$
P(X\le0.10)
=
\int_0^{0.10}
\frac{1}{B(2,18)}
x(1-x)^{17}\,dx
$$

8. Probability that $0.05 \le X \le 0.15$

$$
P(0.05\le X\le0.15)
=
\int_{0.05}^{0.15}
\frac{1}{B(2,18)}
x(1-x)^{17}\,dx
$$

9. Why Beta Distribution is Reasonable

The Beta distribution is appropriate because:

- the variable represents a proportion;
- proportions must lie between 0 and 1;
- the Beta distribution can model many different shapes depending on the parameters;
- it is commonly used for probabilities and quality-control proportions.

Since $\alpha=2$ and $\beta=18$, the distribution is concentrated near small values, meaning low defect rates are more likely.

10. Interpretation of Expectation and Variance

Expectation

$$
E[X]=0.10
$$

This means the average defective proportion is expected to be about 10%.

Variance

$$
\mathrm{Var}(X)=0.004286
$$

The variance measures how much the defective proportion varies from batch to batch.

A small variance indicates that production quality is relatively stable and consistent.

# Q5) Expected Project Completion Time

A construction company is estimating the time required to install a backup generator at a hospital. Let $X$ denote the number of hours needed to complete the installation. Based on expert judgment, the shortest possible time is $4$ hours, the longest possible time is $10$ hours, and the most likely completion time is $6$ hours. Assume that $X$ follows a Triangular distribution with parameters $a=4$, $b=10$, and $c=6$.

Answer the following:

1. Specify the support of $X$.

2. Write down the probability density function $f_X(x)$.

3. Verify that $f_X$ integrates to $1$ over its support.

4. Compute the expectation $E[X]$.

5. Compute the variance $\mathrm{Var}(X)$.

6. Interpret the values of $a$, $b$, and $c$ in the context of this problem.

7. Explain briefly why a Triangular distribution is a reasonable model in this setting.

8. Compute the probability that the installation takes at most $5$ hours, that is, compute

$$
P(X \leq 5).
$$

9. Compute the probability that the installation takes between $5$ and $8$ hours, that is, compute

$$
P(5 \leq X \leq 8).
$$

10. Briefly explain how this model could be used in project planning or sensitivity analysis.


### Answer

Let $X$ denote the installation time (in hours) for the backup generator.

Assume:
$$
X \sim \text{Triangular}(a=4,\; b=10,\; c=6)
$$

where:
- $a=4$ is the minimum possible time,
- $b=10$ is the maximum possible time,
- $c=6$ is the most likely completion time.

1. Support of $X$

The random variable can only take values between 4 and 10 hours.

Thus, the support is:
$$
4 \le X \le 10
$$

2. Probability Density Function

For a triangular distribution:

$$
f_X(x)=
\begin{cases}
\dfrac{2(x-a)}{(b-a)(c-a)}, & a\le x\le c \\[10pt]
\dfrac{2(b-x)}{(b-a)(b-c)}, & c<x\le b \\[10pt]
0, & \text{otherwise}
\end{cases}
$$

Substituting:
$$
a=4,\quad b=10,\quad c=6
$$

gives:

$$
f_X(x)=
\begin{cases}
\dfrac{x-4}{6}, & 4\le x\le 6 \\[10pt]
\dfrac{10-x}{12}, & 6<x\le10 \\[10pt]
0, & \text{otherwise}
\end{cases}
$$

3. Verification that $f_X$ Integrates to 1

We compute:

$$
\int_4^6 \frac{x-4}{6}\,dx
+
\int_6^{10} \frac{10-x}{12}\,dx
$$

First integral:

$$
\int_4^6 \frac{x-4}{6}\,dx
=
\frac{1}{6}\left[\frac{(x-4)^2}{2}\right]_4^6
$$

$$
=
\frac{1}{6}\times \frac{4}{2}
=
\frac{1}{3}
$$

Second integral:

$$
\int_6^{10}\frac{10-x}{12}\,dx
=
\frac{1}{12}\left[10x-\frac{x^2}{2}\right]_6^{10}
$$

$$
=
\frac{2}{3}
$$

Therefore:

$$
\frac{1}{3}+\frac{2}{3}=1
$$

Hence, $f_X$ is a valid probability density function.

4. Expectation of $X$

For a triangular distribution:

$$
E[X]=\frac{a+b+c}{3}
$$

Substituting values:

$$
E[X]=\frac{4+10+6}{3}
$$

$$
E[X]=\frac{20}{3}
$$

$$
E[X]\approx 6.67
$$

hours.

5. Variance of $X$

For a triangular distribution:

$$
\mathrm{Var}(X)
=
\frac{
a^2+b^2+c^2-ab-ac-bc
}{18}
$$

Substituting values:

$$
=
\frac{
16+100+36-40-24-60
}{18}
$$

$$
=
\frac{28}{18}
$$

$$
\mathrm{Var}(X)=\frac{14}{9}
\approx1.56
$$

6. Interpretation of Parameters

- $a=4$: shortest possible installation time
- $b=10$: longest possible installation time
- $c=6$: most likely installation time

7. Why Triangular Distribution is Reasonable

The triangular distribution is useful when:

- only limited information is available,
- experts can estimate minimum, maximum, and most likely values,
- detailed historical data is unavailable.

It is commonly used in engineering and project management.

8. Probability that Installation Takes at Most 5 Hours

Since $5<6$, use the first part of the density:

$$
P(X\le5)
=
\int_4^5 \frac{x-4}{6}\,dx
$$

$$
=
\frac{1}{6}\left[\frac{(x-4)^2}{2}\right]_4^5
$$

$$
=
\frac{1}{12}

$$
P(X\le5)\approx0.0833
$$

9. Probability that Installation Takes Between 5 and 8 Hours

$$
P(5\le X\le8)
=
\int_5^6 \frac{x-4}{6}\,dx
+
\int_6^8 \frac{10-x}{12}\,dx
$$

First integral:

$$
=
\frac{1}{6}\left[\frac{(x-4)^2}{2}\right]_5^6

$$

#Q6) Expected Project Completion Time Using PERT

Based on expert judgment, the shortest possible time is $a=8$ days, the longest possible time is $b=20$ days, and the most likely completion time is $c=11$ days. Assume that $X$ follows a PERT distribution with parameters $(a,b,c)$.

The PERT distribution is defined as a Beta distribution scaled to the interval $[a,b]$, with internal shape parameters

$$
\alpha = 1 + 4\left( \frac{c-a}{b-a} \right)
$$

and

$$
\beta = 1 + 4\left( \frac{b-c}{b-a} \right).
$$

Its probability density function is

$$
f(x)=\frac{(x-a)^{\alpha-1}(b-x)^{\beta-1}}{B(\alpha,\beta)(b-a)^{\alpha+\beta-1}}, \qquad a \leq x \leq b,
$$

where

$$
B(\alpha,\beta)=\int_0^1 t^{\alpha-1}(1-t)^{\beta-1},dt.
$$

Answer the following:

1. Compute the values of the shape parameters $\alpha$ and $\beta$.

2. Specify the support of $X$.

3. Write down the probability density function $f_X(x)$ for this model.

4. State the probability space $(\Omega,\mathfrak F,P)$ suitable for this continuous model.

### Answer

Let $X$ denote the number of days required to complete the software integration task.

$$

$$

- $a$ = minimum possible time,

$$
\alpha = 1 + 4\left(\frac{c-a}{b-a}\right)

\beta = 1 + 4\left(\frac{b-c}{b-a}\right)

$$
=

$$

$$

#Q7) If the PERT Expectation is to Match Our Intuition

$$
a=5
$$

$$
c=11

E[X]=c

$$

11=\frac{5+4(11)+b}{6}

Multiply both sides by 6:

66=5+44+b

b=17
$$

Answer for Maximum Time

$$
\boxed{b=17}

Variance

$$
\mathrm{Var}(X)
=
\frac{(E[X]-a)(b-E[X])}{7}

Substitute values:

$$
=
