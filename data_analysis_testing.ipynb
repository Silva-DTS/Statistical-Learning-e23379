##Custom Plotting with PlottingMethods

```python
from data_analysis.core import PlottingMethods

PLT = PlottingMethods()
```

```python
# See all available plotting methods
response = PLT.get_methods_info()
pd.DataFrame(response.get('response', {}))
```

```python
inspector.df.columns
```

```python
result = PLT.plot_pie_chart(names='Sex', values='Fare', data=inspector.df)
print(result.get('status'))
PLT.display_image(result=result)
```

```python
result = PLT.plot_bar_chart(x='Sex', y='Fare', data=inspector.df)
PLT.display_image(result=result)
```

```python
result = PLT.plot_histogram(x='Fare', data=inspector.df)
PLT.display_image(result=result)
```

```python
title = 'Survival Rate'
result = PLT.plot_heat_map(
    values='Survived',
    index='Sex',
    columns='Pclass',
    aggregade_method='mean',
    title=title,
    data=inspector.df
)
PLT.display_image(result=result)
```

```python
result = PLT.plot_sankey_diagram(
    source_column='Sex',
    target_column='Embarked',
    values='Fare',
    data=inspector.df
)
print(result.get('status'))
PLT.display_image(result=result)
```

```python
title = 'Passenger Fare by Class and Sex'
result = PLT.plot_simple_sunburst_graph(
    path=['Sex', 'Pclass'],
    values='Fare',
    data=inspector.df,
    title=title
)
print(result.get('status'))
PLT.display_image(result=result)
```
